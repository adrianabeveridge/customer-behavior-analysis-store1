## 📁 4. DADOS (CÓDIGO PYTHON)

```python
users = [
    ['32415', ['mike', 'reed'], 32, ['electronics', 'sport', 'books'], [894, 213, 173]],
    ['31980', ['kate', 'morgan'], 24, ['clothes', 'books'], [439, 390]],
    ['32156', ['john', 'doe'], 37, ['electronics', 'home', 'food'], [459, 120, 99]],
]
```

## 5. RECEITA TOTAL

```python
revenue = 0

for user in users:
    revenue += sum(user[4])

print("Receita total:", revenue)
```

## 6. FILTRO DE CLIENTES (IDADE + GASTO)

```python
for user in users:
    age = user[2]
    total_spent = sum(user[4])

    if age < 30 and total_spent > 1000:
        print("Cliente:", user[1][0])
```

## 7. CLIENTES COM COMPRA DE ROUPAS

```python
for user in users:
    if 'clothes' in user[3]:
        print(f"{user[1][0]} tem {user[2]} anos")
```
